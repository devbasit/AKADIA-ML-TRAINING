## Machine Learning Tasks

### Task 1: Regression (Predicting Exam Score)

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error, r2_score
import joblib

# Ensure no missing values in the relevant columns for regression (features and target)
# Drop rows with any NaN in 'Study_Hours', 'Attendance_Rate', or 'Exam_Score'
# This is a robust way to handle any lingering NaNs that might affect the models.
cleaned_data_reg = data[['Study_Hours', 'Attendance_Rate', 'Exam_Score']].dropna()

# Define features (X) and target (y) for regression from the cleaned data
X_reg = cleaned_data_reg[['Study_Hours', 'Attendance_Rate']]
y_reg = cleaned_data_reg['Exam_Score']

# Split the data into training and testing sets
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(X_reg, y_reg, test_size=0.2, random_state=42)
```

```python
# Diagnostic check for NaNs in regression training/testing sets
print('NaNs in X_train_reg:', X_train_reg.isnull().sum().sum())
print('NaNs in y_train_reg:', y_train_reg.isnull().sum())
print('NaNs in X_test_reg:', X_test_reg.isnull().sum().sum())
print('NaNs in y_test_reg:', y_test_reg.isnull().sum())
```

### Model 1: Linear Regression

```python
# Train a Linear Regression model
linear_model = LinearRegression()
linear_model.fit(X_train_reg, y_train_reg)

# Make predictions
y_pred_linear = linear_model.predict(X_test_reg)

# Evaluate the model
mse_linear = mean_squared_error(y_test_reg, y_pred_linear)
r2_linear = r2_score(y_test_reg, y_pred_linear)

print(f"Linear Regression MSE: {mse_linear:.2f}")
print(f"Linear Regression R2 Score: {r2_linear:.2f}")

# Save the model
joblib.dump(linear_model, 'linear_regression_model.pkl')
```

### Model 2: Decision Tree Regressor

```python
# Train a Decision Tree Regressor model
decision_tree_model = DecisionTreeRegressor(random_state=42)
decision_tree_model.fit(X_train_reg, y_train_reg)

# Make predictions
y_pred_tree = decision_tree_model.predict(X_test_reg)

# Evaluate the model
mse_tree = mean_squared_error(y_test_reg, y_pred_tree)
r2_tree = r2_score(y_test_reg, y_pred_tree)

print(f"Decision Tree Regressor MSE: {mse_tree:.2f}")
print(f"Decision Tree Regressor R2 Score: {r2_tree:.2f}")

# Save the model
joblib.dump(decision_tree_model, 'decision_tree_regressor_model.pkl')
```

### Task 3: Clustering

```python
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
import joblib # Import joblib for saving the scaler

# Select numerical features for clustering
X_cluster = data[['Study_Hours', 'Attendance_Rate', 'Exam_Score']]

# Standardize the features (important for distance-based algorithms like KMeans)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_cluster)
X_scaled_df = pd.DataFrame(X_scaled, columns=X_cluster.columns)

# Save the scaler
joblib.dump(scaler, 'scaler.pkl')

# For clustering, we don't have a target variable (y). We'll find inherent structures.
# We need to decide on the number of clusters. Let's choose 3 for demonstration.
k = 3
```

### Model 1: K-Means Clustering

```python
# Train a K-Means clustering model
kmeans_model = KMeans(n_clusters=k, random_state=42, n_init=10) # n_init to suppress warning
kmeans_model.fit(X_scaled_df)

# Add cluster labels to the original data
data['KMeans_Cluster'] = kmeans_model.labels_

# Evaluate the clustering (e.g., using Silhouette Score)
silhouette_kmeans = silhouette_score(X_scaled_df, kmeans_model.labels_)
print(f"K-Means Silhouette Score: {silhouette_kmeans:.2f}")

# Display the count of samples per cluster
print("\nK-Means Cluster Distribution:")
print(data['KMeans_Cluster'].value_counts())

# Save the model
joblib.dump(kmeans_model, 'kmeans_clustering_model.pkl')
```

### Model 2: Agglomerative Clustering

```python
# Train an Agglomerative Clustering model
agglo_model = AgglomerativeClustering(n_clusters=k)
agglo_model.fit(X_scaled_df)

# Add cluster labels to the original data
data['Agglo_Cluster'] = agglo_model.labels_

# Evaluate the clustering
silhouette_agglo = silhouette_score(X_scaled_df, agglo_model.labels_)
print(f"Agglomerative Clustering Silhouette Score: {silhouette_agglo:.2f}")

# Display the count of samples per cluster
print("\nAgglomerative Cluster Distribution:")
print(data['Agglo_Cluster'].value_counts())

# Save the model
joblib.dump(agglo_model, 'agglomerative_clustering_model.pkl')
```

### Task 2: Classification (Pass/Fail)

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report

# Create a 'Pass/Fail' target variable (assuming a pass score >= 70)
data['Pass_Fail'] = (data['Exam_Score'] >= 70).astype(int)

# Define features (X) and target (y) for classification
# Exclude 'Exam_Score' itself from features as it's directly used to create the target
X_cls = data[['Study_Hours', 'Attendance_Rate', 'Gender', 'Internet_Access']]
y_cls = data['Pass_Fail']

# Convert categorical features to numerical using one-hot encoding
X_cls = pd.get_dummies(X_cls, columns=['Gender', 'Internet_Access'], drop_first=True)

# Split the data into training and testing sets
X_train_cls, X_test_cls, y_train_cls, y_test_cls = train_test_split(X_cls, y_cls, test_size=0.2, random_state=42)
```

### Model 1: Logistic Regression Classifier

```python
# Train a Logistic Regression model
logistic_model = LogisticRegression(random_state=42, solver='liblinear') # liblinear is good for small datasets
logistic_model.fit(X_train_cls, y_train_cls)

# Make predictions
y_pred_logistic = logistic_model.predict(X_test_cls)

# Evaluate the model
accuracy_logistic = accuracy_score(y_test_cls, y_pred_logistic)
report_logistic = classification_report(y_test_cls, y_pred_logistic)

print(f"Logistic Regression Accuracy: {accuracy_logistic:.2f}")
print("Logistic Regression Classification Report:")
print(report_logistic)

# Save the model
joblib.dump(logistic_model, 'logistic_regression_classifier_model.pkl')
```

### Model 2: Decision Tree Classifier

```python
# Train a Decision Tree Classifier model
decision_tree_classifier = DecisionTreeClassifier(random_state=42)
decision_tree_classifier.fit(X_train_cls, y_train_cls)

# Make predictions
y_pred_dtc = decision_tree_classifier.predict(X_test_cls)

# Evaluate the model
accuracy_dtc = accuracy_score(y_test_cls, y_pred_dtc)
report_dtc = classification_report(y_test_cls, y_pred_dtc)

print(f"Decision Tree Classifier Accuracy: {accuracy_dtc:.2f}")
print("Decision Tree Classifier Classification Report:")
print(report_dtc)

# Save the model
joblib.dump(decision_tree_classifier, 'decision_tree_classifier_model.pkl')
```

```python
# Diagnostic check for NaNs in regression training/testing sets
print('NaNs in X_train_reg:', X_train_reg.isnull().sum().sum())
print('NaNs in y_train_reg:', y_train_reg.isnull().sum())
print('NaNs in X_test_reg:', X_test_reg.isnull().sum().sum())
print('NaNs in y_test_reg:', y_test_reg.isnull().sum())
```

## Deployment Scripts

### 1. Streamlit Application (app_streamlit.py)

```python
%%writefile app_streamlit.py
import streamlit as st
import joblib
import pandas as pd

# Load models and scaler
@st.cache_resource
def load_models():
    models = {
        'regression': {
            'Linear Regression': joblib.load('linear_regression_model.pkl'),
            'Decision Tree Regressor': joblib.load('decision_tree_regressor_model.pkl')
        },
        'classification': {
            'Logistic Regression': joblib.load('logistic_regression_classifier_model.pkl'),
            'Decision Tree Classifier': joblib.load('decision_tree_classifier_model.pkl')
        },
        'clustering': {
            'K-Means': joblib.load('kmeans_clustering_model.pkl'),
            'Agglomerative Clustering': joblib.load('agglomerative_clustering_model.pkl')
        }
    }
    scaler = joblib.load('scaler.pkl')
    return models, scaler

models, scaler = load_models()

st.title("ML Model Deployment Test")

# Task Selection
task = st.sidebar.selectbox(
    "Select Task",
    ("Regression", "Classification", "Clustering")
)

# Input Data
st.header("Input Data")

if task == "Regression":
    model_choice = st.sidebar.selectbox(
        "Select Regression Model",
        list(models['regression'].keys())
    )
    study_hours = st.number_input("Study Hours", min_value=0.0, max_value=20.0, value=5.0)
    attendance_rate = st.number_input("Attendance Rate (%)", min_value=0.0, max_value=100.0, value=80.0)
    input_df = pd.DataFrame([{'Study_Hours': study_hours, 'Attendance_Rate': attendance_rate}])

    if st.button("Predict Exam Score"):
        model = models['regression'][model_choice]
        prediction = model.predict(input_df)[0]
        st.success(f"Predicted Exam Score ({model_choice}): {prediction:.2f}")

elif task == "Classification":
    model_choice = st.sidebar.selectbox(
        "Select Classification Model",
        list(models['classification'].keys())
    )
    study_hours = st.number_input("Study Hours", min_value=0.0, max_value=20.0, value=5.0)
    attendance_rate = st.number_input("Attendance Rate (%)", min_value=0.0, max_value=100.0, value=80.0)
    gender = st.selectbox("Gender", ("Male", "Female"))
    internet_access = st.selectbox("Internet Access", ("Yes", "No"))

    input_data = {
        'Study_Hours': study_hours,
        'Attendance_Rate': attendance_rate,
        'Gender_Male': 1 if gender == 'Male' else 0,
        'Internet_Access_Yes': 1 if internet_access == 'Yes' else 0
    }
    input_df = pd.DataFrame([input_data])

    # Ensure column order matches training data for models that are sensitive to it
    # For simplicity, assuming the order created by get_dummies was: Study_Hours, Attendance_Rate, Gender_Male, Internet_Access_Yes
    # In a real app, you might want to save the feature columns order.
    feature_cols = ['Study_Hours', 'Attendance_Rate', 'Gender_Male', 'Internet_Access_Yes']
    input_df = input_df[feature_cols]

    if st.button("Predict Pass/Fail"):
        model = models['classification'][model_choice]
        prediction = model.predict(input_df)[0]
        result = "Pass" if prediction == 1 else "Fail"
        st.success(f"Prediction ({model_choice}): {result}")

elif task == "Clustering":
    model_choice = st.sidebar.selectbox(
        "Select Clustering Model",
        list(models['clustering'].keys())
    )
    study_hours = st.number_input("Study Hours", min_value=0.0, max_value=20.0, value=5.0)
    attendance_rate = st.number_input("Attendance Rate (%)", min_value=0.0, max_value=100.0, value=80.0)
    exam_score = st.number_input("Exam Score", min_value=0.0, max_value=100.0, value=75.0)

    input_df = pd.DataFrame([{'Study_Hours': study_hours, 'Attendance_Rate': attendance_rate, 'Exam_Score': exam_score}])

    if st.button("Assign Cluster"):
        model = models['clustering'][model_choice]
        # Scale the input data using the loaded scaler
        scaled_input = scaler.transform(input_df[['Study_Hours', 'Attendance_Rate', 'Exam_Score']])
        cluster_assignment = model.predict(scaled_input)[0]
        st.success(f"Assigned Cluster ({model_choice}): {cluster_assignment}")
```

### 2. FastAPI Application (app_fastapi.py)

```python
%%writefile app_fastapi.py
from fastapi import FastAPI, HTTPException
from pydantic import BaseModel
import joblib
import pandas as pd
from typing import Literal

app = FastAPI(title="ML Model Deployment API")

# Load models and scaler
# In a real application, you might want to load these once at startup
# and store them globally or in app.state.
regression_models = {
    'linear_regression': joblib.load('linear_regression_model.pkl'),
    'decision_tree_regressor': joblib.load('decision_tree_regressor_model.pkl')
}
classification_models = {
    'logistic_regression': joblib.load('logistic_regression_classifier_model.pkl'),
    'decision_tree_classifier': joblib.load('decision_tree_classifier_model.pkl')
}
clustering_models = {
    'kmeans': joblib.load('kmeans_clustering_model.pkl'),
    'agglomerative_clustering': joblib.load('agglomerative_clustering_model.pkl')
}
scaler = joblib.load('scaler.pkl')

# Pydantic models for request bodies
class RegressionInput(BaseModel):
    study_hours: float
    attendance_rate: float
    model_name: Literal["linear_regression", "decision_tree_regressor"]

class ClassificationInput(BaseModel):
    study_hours: float
    attendance_rate: float
    gender: Literal["Male", "Female"]
    internet_access: Literal["Yes", "No"]
    model_name: Literal["logistic_regression", "decision_tree_classifier"]

class ClusteringInput(BaseModel):
    study_hours: float
    attendance_rate: float
    exam_score: float
    model_name: Literal["kmeans", "agglomerative_clustering"]

@app.post("/predict/regression")
async def predict_regression(input_data: RegressionInput):
    model = regression_models.get(input_data.model_name)
    if not model:
        raise HTTPException(status_code=400, detail="Invalid regression model name")

    df = pd.DataFrame([{
        'Study_Hours': input_data.study_hours,
        'Attendance_Rate': input_data.attendance_rate
    }])

    prediction = model.predict(df)[0]
    return {"task": "regression", "model": input_data.model_name, "predicted_exam_score": round(prediction, 2)}

@app.post("/predict/classification")
async def predict_classification(input_data: ClassificationInput):
    model = classification_models.get(input_data.model_name)
    if not model:
        raise HTTPException(status_code=400, detail="Invalid classification model name")

    # Prepare input data for one-hot encoding
    data_dict = {
        'Study_Hours': [input_data.study_hours],
        'Attendance_Rate': [input_data.attendance_rate],
        'Gender': [input_data.gender],
        'Internet_Access': [input_data.internet_access]
    }
    df = pd.DataFrame(data_dict)

    # Apply one-hot encoding consistent with training
    # Assuming columns were 'Gender_Male', 'Internet_Access_Yes'
    df['Gender_Male'] = df['Gender'].apply(lambda x: 1 if x == 'Male' else 0)
    df['Internet_Access_Yes'] = df['Internet_Access'].apply(lambda x: 1 if x == 'Yes' else 0)

    # Select and order features as trained
    feature_cols = ['Study_Hours', 'Attendance_Rate', 'Gender_Male', 'Internet_Access_Yes']
    input_features = df[feature_cols]

    prediction = model.predict(input_features)[0]
    result = "Pass" if prediction == 1 else "Fail"
    return {"task": "classification", "model": input_data.model_name, "predicted_pass_fail": result}

@app.post("/predict/clustering")
async def predict_clustering(input_data: ClusteringInput):
    model = clustering_models.get(input_data.model_name)
    if not model:
        raise HTTPException(status_code=400, detail="Invalid clustering model name")

    df = pd.DataFrame([{
        'Study_Hours': input_data.study_hours,
        'Attendance_Rate': input_data.attendance_rate,
        'Exam_Score': input_data.exam_score
    }])

    # Scale the input data
    scaled_input = scaler.transform(df[['Study_Hours', 'Attendance_Rate', 'Exam_Score']])

    cluster_assignment = model.predict(scaled_input)[0]
    return {"task": "clustering", "model": input_data.model_name, "assigned_cluster": int(cluster_assignment)}
```

### 3. Flask Application (app_flask.py)

```python
%%writefile app_flask.py
from flask import Flask, request, jsonify
import joblib
import pandas as pd

app = Flask(__name__)

# Load models and scaler
# In a real application, consider using a global variable or app context for efficiency
regression_models = {
    'linear_regression': joblib.load('linear_regression_model.pkl'),
    'decision_tree_regressor': joblib.load('decision_tree_regressor_model.pkl')
}
classification_models = {
    'logistic_regression': joblib.load('logistic_regression_classifier_model.pkl'),
    'decision_tree_classifier': joblib.load('decision_tree_classifier_model.pkl')
}
clustering_models = {
    'kmeans': joblib.load('kmeans_clustering_model.pkl'),
    'agglomerative_clustering': joblib.load('agglomerative_clustering_model.pkl')
}
scaler = joblib.load('scaler.pkl')

@app.route('/predict/regression', methods=['POST'])
def predict_regression_flask():
    data = request.get_json()
    study_hours = data.get('study_hours')
    attendance_rate = data.get('attendance_rate')
    model_name = data.get('model_name')

    if not all([study_hours, attendance_rate, model_name]):
        return jsonify({"error": "Missing data points or model name"}), 400

    model = regression_models.get(model_name)
    if not model:
        return jsonify({"error": "Invalid regression model name"}), 400

    df = pd.DataFrame([{
        'Study_Hours': study_hours,
        'Attendance_Rate': attendance_rate
    }])

    prediction = model.predict(df)[0]
    return jsonify({"task": "regression", "model": model_name, "predicted_exam_score": round(prediction, 2)})

@app.route('/predict/classification', methods=['POST'])
def predict_classification_flask():
    data = request.get_json()
    study_hours = data.get('study_hours')
    attendance_rate = data.get('attendance_rate')
    gender = data.get('gender')
    internet_access = data.get('internet_access')
    model_name = data.get('model_name')

    if not all([study_hours, attendance_rate, gender, internet_access, model_name]):
        return jsonify({"error": "Missing data points or model name"}), 400

    model = classification_models.get(model_name)
    if not model:
        return jsonify({"error": "Invalid classification model name"}), 400

    data_dict = {
        'Study_Hours': [study_hours],
        'Attendance_Rate': [attendance_rate],
        'Gender': [gender],
        'Internet_Access': [internet_access]
    }
    df = pd.DataFrame(data_dict)

    df['Gender_Male'] = df['Gender'].apply(lambda x: 1 if x == 'Male' else 0)
    df['Internet_Access_Yes'] = df['Internet_Access'].apply(lambda x: 1 if x == 'Yes' else 0)

    feature_cols = ['Study_Hours', 'Attendance_Rate', 'Gender_Male', 'Internet_Access_Yes']
    input_features = df[feature_cols]

    prediction = model.predict(input_features)[0]
    result = "Pass" if prediction == 1 else "Fail"
    return jsonify({"task": "classification", "model": model_name, "predicted_pass_fail": result})

@app.route('/predict/clustering', methods=['POST'])
def predict_clustering_flask():
    data = request.get_json()
    study_hours = data.get('study_hours')
    attendance_rate = data.get('attendance_rate')
    exam_score = data.get('exam_score')
    model_name = data.get('model_name')

    if not all([study_hours, attendance_rate, exam_score, model_name]):
        return jsonify({"error": "Missing data points or model name"}), 400

    model = clustering_models.get(model_name)
    if not model:
        return jsonify({"error": "Invalid clustering model name"}), 400

    df = pd.DataFrame([{
        'Study_Hours': study_hours,
        'Attendance_Rate': attendance_rate,
        'Exam_Score': exam_score
    }])

    scaled_input = scaler.transform(df[['Study_Hours', 'Attendance_Rate', 'Exam_Score']])

    cluster_assignment = model.predict(scaled_input)[0]
    return jsonify({"task": "clustering", "model": model_name, "assigned_cluster": int(cluster_assignment)})

if __name__ == '__main__':
    app.run(debug=True, host='0.0.0.0', port=5000)
```
