```python
# Drop rows with any null values
df_cleaned_drop = df.dropna()
display(df_cleaned_drop.head())
df_cleaned_drop.isnull().sum()
```

```python
# Fill null values in 'Study_Hours' with the mean of the column
df_filled = df.copy()
df_filled['Study_Hours'] = df_filled['Study_Hours'].fillna(df_filled['Study_Hours'].min())
df_filled['Attendance_Rate'] = df_filled['Attendance_Rate'].fillna(df_filled['Attendance_Rate'].mean())
display(df_filled.head())
df_filled.isnull().sum()
```

## Grouping and Aggregation

```python
# Group by 'State' and calculate the mean 'Exam_Score'
state_exam_score = df_filled.groupby('State')['Exam_Score'].mean().reset_index()
display(state_exam_score)
```

```python
# Group by 'Gender' and 'State' and calculate multiple aggregations
gender_state_agg = df_filled.groupby(['Gender', 'State']).agg(
    mean_exam_score=('Exam_Score', 'mean'),
    total_students=('Student_ID', 'count'),
    avg_study_hours=('Study_Hours', 'mean')
).reset_index()
display(gender_state_agg)
```

## Merging DataFrames

Merging (or joining) DataFrames is crucial when you have related data spread across multiple tables. We'll create a dummy DataFrame for state populations and merge it with our `state_exam_score` DataFrame.

```python
# Create a dummy DataFrame for state populations
state_population_data = {
    'State': ['Abuja', 'Oyo', 'Ogun', 'Lagos', 'Anambra', 'Enugu', 'Delta', 'Kaduna', 'Kano', 'Rivers'],
    'Population': [4000000, 7000000, 6000000, 20000000, 5000000, 4500000, 4000000, 7500000, 15000000, 7000000]
}
state_population_df = pd.DataFrame(state_population_data)
display(state_population_df.head())

# Merge state_exam_score with state_population_df
# We'll use an inner merge to keep only matching states
merged_df = pd.merge(state_exam_score, state_population_df, on='State', how='inner')
display(merged_df)
```

## Applying Functions

The `apply()` method is very powerful for applying a function along an axis of the DataFrame (rows or columns) or to a Series. Let's create a new column 'Grade_Category' based on 'Exam_Score'.

```python
# Define a function to categorize exam scores
def categorize_score(score):
    if score >= 90:
        return 'Excellent'
    elif score >= 70:
        return 'Good'
    elif score >= 50:
        return 'Average'
    else:
        return 'Needs Improvement'

# Apply the function to the 'Exam_Score' column to create 'Grade_Category'
df_filled['Grade_Category'] = df_filled['Exam_Score'].apply(categorize_score)
display(df_filled[['Exam_Score', 'Grade_Category']].head())
display(df_filled['Grade_Category'].value_counts())
```

## Working with Categorical Data

Categorical data type in pandas is useful for columns that contain a limited number of unique values. It can save memory and speed up certain operations. Let's convert the 'State' and 'Gender' columns to categorical types.

```python
# Convert 'State' and 'Gender' columns to categorical type
df_filled['State'] = df_filled['State'].astype('category')
df_filled['Gender'] = df_filled['Gender'].astype('category')

# Check the data types after conversion
display(df_filled.info())
```

## Handling Duplicates

Duplicate data can skew your analysis. Pandas provides convenient methods to identify and remove duplicate rows based on all or selected columns.

```python
# Create a DataFrame with some explicit duplicates for demonstration
data_with_duplicates = {
    'ID': [1, 2, 3, 1, 4, 5, 3],
    'Value': ['A', 'B', 'C', 'A', 'D', 'E', 'C']
}
df_dup = pd.DataFrame(data_with_duplicates)
display("Original DataFrame with Duplicates:", df_dup)

# Identify duplicate rows (returns a boolean Series)
duplicates = df_dup.duplicated()
display("Boolean Series for Duplicates:", duplicates)
display("Number of Duplicate Rows:", duplicates.sum())

# Drop duplicate rows (keeping the first occurrence by default)
df_no_dup = df_dup.drop_duplicates()
display("DataFrame after dropping duplicates:", df_no_dup)

# To demonstrate dropping duplicates in our main DataFrame, let's check if there are any
# (Based on Student_ID, there shouldn't be, but it's good practice)
display("Duplicates in df_filled (based on all columns):")
display(df_filled.duplicated().sum())

# If we wanted to check for duplicates in a subset of columns, for example, State and Gender
display("Duplicates in df_filled (based on 'State' and 'Gender' only):")
display(df_filled.duplicated(subset=['State', 'Gender']).sum())
```

## Pivot Tables

Pivot tables are an excellent way to summarize and reorganize (pivot) data from a DataFrame. You can specify `index` (rows), `columns` (columns), `values` (data to aggregate), and `aggfunc` (aggregation function).

```python
# Create a pivot table showing the mean Exam_Score by Gender and State
pivot_table_exam_score = pd.pivot_table(df_filled,
                                      values='Exam_Score',
                                      index='State',
                                      columns='Gender',
                                      aggfunc='mean')
display("Pivot Table: Mean Exam Score by State and Gender:", pivot_table_exam_score)

# Another example: count of students by Internet_Access and Grade_Category
pivot_table_counts = pd.pivot_table(df_filled,
                                    values='Student_ID',
                                    index='Grade_Category',
                                    columns='Internet_Access',
                                    aggfunc='count')
display("Pivot Table: Count of Students by Grade Category and Internet Access:", pivot_table_counts)
```

## Data Reshaping with `melt`

The `melt()` function is used to transform or 'unpivot' DataFrames from a wide format to a long format. It's particularly useful when you want to convert columns into rows, which is often required for certain types of analysis or plotting (e.g., using Seaborn).

```python
# Let's melt a subset of df_filled to demonstrate
# We'll keep 'Student_ID' as identifier variables (id_vars)
# and melt 'Study_Hours', 'Attendance_Rate', and 'Exam_Score' into 'variable' and 'value' columns
melted_df = df_filled.melt(id_vars=['Student_ID', 'State', 'Gender'],
                           value_vars=['Study_Hours', 'Attendance_Rate', 'Exam_Score'],
                           var_name='Metric',
                           value_name='Value')
display("Original df_filled head:", df_filled.head())
display("Melted DataFrame head:", melted_df.head())
display("Shape of original df_filled:", df_filled.shape)
display("Shape of melted_df:", melted_df.shape)
```

## String Manipulation

Pandas provides a rich set of string methods that are accessible via the `.str` accessor on Series containing string data. These methods make it easy to perform operations like converting case, checking for substrings, splitting, and replacing.

```python
# Let's assume the 'State' column might need some cleaning or transformation
# Convert 'State' names to uppercase
df_filled['State_Upper'] = df_filled['State'].str.upper()
display("Original State and new State_Upper columns:", df_filled[['State', 'State_Upper']].head())

# Check if a state name contains 'A' (e.g., 'Abuja', 'Anambra')
df_filled['State_Contains_A'] = df_filled['State'].str.contains('a', case=False)
display("States containing 'a' (case-insensitive):", df_filled[['State', 'State_Contains_A']].value_counts().sort_index())

# Replace a part of the string (e.g., replace 'Abuja' with 'FCT Abuja')
df_filled['State_Renamed'] = df_filled['State'].str.replace('Abuja', 'FCT Abuja')
display("States after renaming Abuja:", df_filled[['State', 'State_Renamed']].head())
```

## MultiIndex (Hierarchical Indexing)

MultiIndex objects are useful for representing data with multiple levels of indexing. This allows for more complex data organization and sophisticated data analysis, especially for data that naturally has a hierarchical relationship.

```python
# Let's create a DataFrame with a MultiIndex from our gender_state_agg
# The gender_state_agg already has 'Gender' and 'State' that can form a MultiIndex
multi_indexed_df = gender_state_agg.set_index(['Gender', 'State'])
display("DataFrame with MultiIndex:", multi_indexed_df)

# Access data using MultiIndex
display("Accessing data for Female students in Abuja:", multi_indexed_df.loc[('Female', 'Abuja')])
display("Accessing data for all Male students:", multi_indexed_df.loc['Male'])
```

## Plotting with Pandas (and Matplotlib/Seaborn)

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Bar plot of average exam score by state
plt.figure(figsize=(10, 6))
sns.barplot(x='State', y='Exam_Score', data=state_exam_score)
plt.title('Average Exam Score by State')
plt.ylabel('Average Exam Score')
plt.xlabel('State')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Histogram of Exam Scores
plt.figure(figsize=(8, 5))
sns.histplot(df_filled['Exam_Score'], kde=True, bins=15)
plt.title('Distribution of Exam Scores')
plt.xlabel('Exam Score')
plt.ylabel('Frequency')
plt.show()
```

```python
# Box plot of Exam Score by Gender
plt.figure(figsize=(8, 5))
sns.boxplot(x='Gender', y='Exam_Score', data=df_filled)
plt.title('Exam Score Distribution by Gender')
plt.xlabel('Gender')
plt.ylabel('Exam Score')
plt.show()
```
